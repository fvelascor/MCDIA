# **CHALLENGE: LENGUAJE DE PROGRAMACIÓN PYTHON**

Una vez visto el ***Hands - On: Lenguaje de programación python***, se presentan la siguientes actividades para que el alumnado pueda repasar y reforzar lo aprendido dentro de la clase.

Dentro de la siguiente dinámica, **se creará una agenda de cinco contactos** haciendo uso de las **estructuras de datos** utilizando **Python**.

Donde los datos que debe contener cada contacto de la angenda son:

*   **Nombre**
*   **Teléfono**
*   **Correo**

# --------------------------------------------------------
# PASO 1: CREACIÓN DE LA AGENDA
 La agenda es una LISTA que contiene 5 DICCIONARIOS.
 Cada diccionario representa un contacto con:
 nombre, teléfono y correo.
# --------------------------------------------------------

```python
agenda = [
    {
        'nombre': 'Ana García',
        'teléfono': '555-1001',
        'correo': 'ana.garcia@email.com'
    },
    {
        'nombre': 'Carlos Pérez',
        'teléfono': '555-2002',
        'correo': 'carlos.perez@email.com'
    },
    {
        'nombre': 'María López',
        'teléfono': '555-3003',
        'correo': 'maria.lopez@email.com'
    },
    {
        'nombre': 'Fernando Torres',
        'teléfono': '555-4004',
        'correo': 'fernando.torres@email.com'
    },
    {
        'nombre': 'Lucía Martínez',
        'teléfono': '555-5005',
        'correo': 'lucia.martinez@email.com'
    }
]

print("✅ Agenda creada con", len(agenda), "contactos.\n")
```

# ------------------------------------------------------------
# PASO 2: LECTURA — Mostrar todos los contactos
 Se recorre la lista con un índice para mostrar cada contacto.
# ------------------------------------------------------------


```python
print("=" * 40)
print("         📋 AGENDA DE CONTACTOS")
print("=" * 40)

for i in range(len(agenda)):
    print(f"\nContacto #{i + 1}")
    print("  Nombre   :", agenda[i]['nombre'])
    print("  Teléfono :", agenda[i]['teléfono'])
    print("  Correo   :", agenda[i]['correo'])

print("\n" + "=" * 40)
```

# ------------------------------------------------------------
# PASO 3: LECTURA — Acceder a un contacto específico
 Indexación: accedemos al tercer contacto (índice 2).
# -

```python
print("\n🔍 Accediendo al contacto en el índice 2 (María López):")
print(agenda[2])

# Acceso a un dato específico dentro del contacto
print("\n📧 Correo de María López:", agenda[2]['correo'])
```

# --------------------------------------------------------
# PASO 4: ACTUALIZACIÓN — Modificar un dato existente
 Cambiamos el teléfono de Carlos Pérez (índice 1).
# --------------------------------------------------------

```python
print("\n✏️  Actualizando teléfono de Carlos Pérez...")
agenda[1]['teléfono'] = '555-9999'
print("  Nuevo teléfono:", agenda[1]['teléfono'])
```

# --------------------------------------------------------
# PASO 5: INSERCIÓN — Agregar un nuevo contacto
 Usamos .append() para agregar al final de la lista.
# --------------------------------------------------------

```python
nuevo_contacto = {
    'nombre': 'Roberto Sánchez',
    'teléfono': '555-6006',
    'correo': 'roberto.sanchez@email.com'
}

agenda.append(nuevo_contacto)
print("\n➕ Contacto agregado:", nuevo_contacto['nombre'])
print("  Total de contactos ahora:", len(agenda))
```

# --------------------------------------------------------
# PASO 6: ELIMINACIÓN — Eliminar un contacto
 Usamos .pop() con el índice del contacto a eliminar.
En este caso eliminat el último contacto agregado.
# --------------------------------------------------------

```python
contacto_eliminado = agenda.pop(-1)
print("\n🗑️  Contacto eliminado:", contacto_eliminado['nombre'])
print("  Total de contactos ahora:", len(agenda))

```

# --------------------------------------------------------
# RESULTADO FINAL — Estado final de la agenda
# --------------------------------------------------------

```python
print("\n" + "=" * 40)
print("     📋 ESTADO FINAL DE LA AGENDA")
print("=" * 40)

for i in range(len(agenda)):
    print(f"\nContacto #{i + 1}")
    print("  Nombre   :", agenda[i]['nombre'])
    print("  Teléfono :", agenda[i]['teléfono'])
    print("  Correo   :", agenda[i]['correo'])

print("\n" + "=" * 40)
print("✅ Challenge completado exitosamente.")
```
