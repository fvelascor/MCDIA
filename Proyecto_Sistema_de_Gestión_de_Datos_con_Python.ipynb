# CELDA 2: Función menú de opciones

```python
def menu_opciones():
    print("\n" + "=" * 45)
    print("       🎬 SISTEMA DE GESTIÓN DE PELÍCULAS")
    print("=" * 45)
    print("  1. Ver todas las películas")
    print("  2. Crear nueva película")
    print("  3. Consultar película")
    print("  4. Actualizar película")
    print("  5. Eliminar película")
    print("  6. Salir del menú")
    print("=" * 45)
```

# ============================================================
# CELDA 3: Función READ — Ver todas las películas
# Recorre la lista con un ciclo for y muestra cada registro
# numerado según su posición en la lista
# ============================================================


```python
def ver_peliculas():
    # Validar si existen registros antes de mostrarlos
    if len(peliculas) == 0:
        print("\n  ⚠️  No hay películas registradas.")
        return

    print("\n" + "=" * 55)
    print("              🎬 CATÁLOGO DE PELÍCULAS")
    print("=" * 55)

    # Ciclo for para recorrer e imprimir cada película
    for i in range(len(peliculas)):
        print(f"\n  Película #{i + 1}")
        print(f"    Título         : {peliculas[i]['titulo']}")
        print(f"    Género         : {peliculas[i]['genero']}")
        print(f"    Año            : {peliculas[i]['año']}")
        print(f"    Calificación   : {peliculas[i]['calificacion']} / 10")

    print("\n" + "=" * 55)
    print(f"  Total de películas: {len(peliculas)}")
    print("=" * 55)
```

# ============================================================
# CELDA 4: Función CREATE — Crear nueva película
# Solicita los datos al usuario, los guarda en un diccionario
# y lo agrega a la lista con .append()
# ============================================================


```python
def crear_pelicula():
    print("\n➕ CREAR NUEVA PELÍCULA")

    # Capturar datos del usuario con input()
    titulo        = input("  Ingresa el título        : ")
    genero        = input("  Ingresa el género        : ")

    # Convertir año a entero y calificación a flotante
    año           = int(input("  Ingresa el año           : "))
    calificacion  = float(input("  Ingresa la calificación  : "))

    # Crear diccionario con los datos ingresados
    nueva_pelicula = {
        'titulo':        titulo,
        'genero':        genero,
        'año':           año,
        'calificacion':  calificacion
    }

    # Agregar el nuevo registro al final de la lista
    peliculas.append(nueva_pelicula)
    print(f"\n  ✅ Película '{titulo}' creada con éxito.")
```

# ============================================================
# CELDA 5: Función READ — Consultar película por título
# Usa un ciclo for para buscar el título ingresado
# Si lo encuentra imprime el registro, si no, notifica
# ============================================================


```python
def consultar_pelicula():
    print("\n🔍 CONSULTAR PELÍCULA")

    busqueda   = input("  Ingresa el título a consultar: ")
    encontrado = False

    # Ciclo for para recorrer la lista en búsqueda del título
    for pelicula in peliculas:
        if pelicula['titulo'].lower() == busqueda.lower():
            print("\n  Película encontrada:")
            print(f"    Título         : {pelicula['titulo']}")
            print(f"    Género         : {pelicula['genero']}")
            print(f"    Año            : {pelicula['año']}")
            print(f"    Calificación   : {pelicula['calificacion']} / 10")
            encontrado = True
            break  # Detener búsqueda al encontrar el registro

    # Si el ciclo no encontró coincidencia
    if not encontrado:
        print("  ❌ Película no encontrada.")
```

# ============================================================
# CELDA 6a: Función auxiliar — Menú de actualización
# Muestra las opciones disponibles para modificar un registro
# ============================================================

```python
def menu_actualizar():
    print("\n  Menú actualizar película:")
    print("    1. Actualizar título")
    print("    2. Actualizar género")
    print("    3. Actualizar año")
    print("    4. Actualizar calificación")
    print("    5. Salir del menú")
```

# ============================================================
# CELDA 6b: Función auxiliar — Buscador de película
# Recorre la lista y retorna el diccionario si lo encuentra
# Retorna None si no existe (comportamiento por defecto)
# ============================================================

```python
def buscar_pelicula(busqueda):
    for pelicula in peliculas:
        if pelicula['titulo'].lower() == busqueda.lower():
            return pelicula  # Retorna el diccionario encontrado
    return None              # Retorna None si no hay coincidencia
```

# ============================================================
# CELDA 6c: Funciones de actualización por campo
# Cada función modifica un único campo del registro
# y muestra el estado actualizado
# ============================================================

```python
def actualizar_titulo(pelicula):
    pelicula['titulo'] = input("    Nuevo título: ")
    print(f"\n  Película actualizada: {pelicula}")

def actualizar_genero(pelicula):
    pelicula['genero'] = input("    Nuevo género: ")
    print(f"\n  Película actualizada: {pelicula}")

def actualizar_año(pelicula):
    pelicula['año'] = int(input("    Nuevo año: "))
    print(f"\n  Película actualizada: {pelicula}")

def actualizar_calificacion(pelicula):
    pelicula['calificacion'] = float(input("    Nueva calificación: "))
    print(f"\n  Película actualizada: {pelicula}")
```

# ============================================================
# CELDA 6d: Función UPDATE — Actualizar película (integradora)
# Busca el registro, muestra el submenú con while y delega
# cada modificación a su función correspondiente
# ============================================================

```python
def actualizar_pelicula():
    print("\n✏️  ACTUALIZAR PELÍCULA")

    busqueda = input("  Ingresa el título de la película a actualizar: ")

    # Usar función buscadora auxiliar
    pelicula = buscar_pelicula(busqueda)

    # Verificar si se encontró el registro
    if pelicula is None:
        print("  ❌ Película no encontrada.")
        return

    # Mostrar el registro encontrado
    print("\n  Película encontrada:")
    print(f"    Título         : {pelicula['titulo']}")
    print(f"    Género         : {pelicula['genero']}")
    print(f"    Año            : {pelicula['año']}")
    print(f"    Calificación   : {pelicula['calificacion']} / 10")

    # Submenú de actualización con ciclo while
    while True:
        menu_actualizar()
        opcion = input("\n  Selecciona una opción: ")

        if opcion == '1':
            actualizar_titulo(pelicula)
        elif opcion == '2':
            actualizar_genero(pelicula)
        elif opcion == '3':
            actualizar_año(pelicula)
        elif opcion == '4':
            actualizar_calificacion(pelicula)
        elif opcion == '5':
            print("  👋 Saliendo del menú de actualización.")
            break  # Salir del ciclo while
        else:
            print("  ❌ Opción no encontrada.")
```

# ============================================================
# CELDA 7: Función DELETE — Eliminar película
# Busca el título ingresado con un ciclo for
# Si lo encuentra lo elimina con .remove()
# Si no, notifica al usuario
# ============================================================


```python
def eliminar_pelicula():
    print("\n🗑️  ELIMINAR PELÍCULA")

    busqueda   = input("  Ingresa el título de la película a eliminar: ")
    encontrado = False

    # Ciclo for para buscar la película
    for pelicula in peliculas:
        if pelicula['titulo'].lower() == busqueda.lower():
            peliculas.remove(pelicula)   # Eliminar el registro encontrado
            print(f"  ✅ Película '{pelicula['titulo']}' eliminada con éxito.")
            encontrado = True
            break  # Detener búsqueda tras eliminar

    # Si no se encontró el registro
    if not encontrado:
        print("  ❌ Película no encontrada.")
```

# ============================================================
# CELDA 8: Función principal — Menú CRUD orquestador
# Integra todas las funciones dentro de un ciclo while
# Cada opción invoca su función correspondiente
# La opción 6 ejecuta break para salir del programa
# ============================================================


```python
def menu_crud():
    while True:
        # Mostrar opciones del menú principal
        menu_opciones()
        opcion = input("\n  Selecciona una opción: ")

        if opcion == '1':
            ver_peliculas()        # READ — Ver todos los registros
        elif opcion == '2':
            crear_pelicula()       # CREATE — Agregar nuevo registro
        elif opcion == '3':
            consultar_pelicula()   # READ — Consultar un registro
        elif opcion == '4':
            actualizar_pelicula()  # UPDATE — Modificar un registro
        elif opcion == '5':
            eliminar_pelicula()    # DELETE — Eliminar un registro
        elif opcion == '6':
            print("\n  👋 Saliendo del sistema. ¡Hasta luego!")
            break                  # Salir del ciclo while
        else:
            print("\n  ❌ Opción no encontrada.")
```

# ============================================================
# PUNTO DE ENTRADA
# Se invoca la función principal para iniciar el sistema
# ============================================================

```python
# ============================================================
# PUNTO DE ENTRADA
# Se invoca la función principal para iniciar el sistema
# ============================================================

menu_crud()
```
